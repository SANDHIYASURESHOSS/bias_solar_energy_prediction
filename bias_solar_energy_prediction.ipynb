```python
# Bias in Solar Energy Prediction

inputs = [10, 20, 30, 40]
weights = 2
bias = 5

for x in inputs:
    output = (x * weights) + bias
    print("Input:", x, "Output:", output)
```

```python
inputs = [10, 20, 30, 40]
weight = 2

print("Without Bias:")
for x in inputs:
    print(x * weight)

print("\nWith Bias:")
bias = 5
for x in inputs:
    print((x * weight) + bias)
```

```python
# Solar Energy Prediction - Bias Analysis

x = 4
weight = 2

bias_values = [-3, 0, 2, 5]

for bias in bias_values:
    z = (x * weight) + bias
    print("Bias =", bias, "Pre-activation =", z)
```

```python
# Decision response using bias

x = 4
weight = 2
threshold = 8

for bias in [-3, 0, 2, 5]:
    z = (x * weight) + bias
    decision = "HIGH" if z >= threshold else "LOW"
    print("Bias =", bias, "| z =", z, "| Decision =", decision)
```

```python
import matplotlib.pyplot as plt

bias_values = [-3, 0, 2, 5]
outputs = [(4 * 2) + b for b in bias_values]

plt.plot(bias_values, outputs, marker='o')
plt.xlabel("Bias")
plt.ylabel("Pre-activation Value")
plt.title("Effect of Bias on Neuron Output")
plt.grid(True)
plt.show()
```
